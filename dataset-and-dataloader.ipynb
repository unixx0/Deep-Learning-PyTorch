```python
from sklearn.datasets import make_classification
```

```python
x, y= make_classification(
    n_samples= 10,
    n_redundant= 0,
    n_features= 3,
    random_state= 42,

)
```

```python
# converting array into tensors

import torch
x= torch.from_numpy(x).float()
y= torch.tensor(data= y, dtype= torch.float32)
```

```python
x
```

```python
y
```

```python
from torch.utils.data import Dataset, DataLoader
```

```python
# dataset class is used for fetching data from the memory

class CustomData(Dataset):

    def __init__(self, features, label):     #features is inputs and label is output; loading the data
        self.features= features
        self.label= label


    def __len__(self):
        return self.features.shape[0]     #returns the total number of rows in our data



    def __getitem__(self, index):
        return self.features[index], self.label[index]   # returns respective row of our data
```

```python
data= CustomData(x, y)
```

```python
data[5]
```

```python
len(data)
```

```python
## getting batches of data
dataloader= DataLoader(dataset= data, batch_size= 2, shuffle= True)

for feature, label in dataloader:
    print(feature)
    print(label)
    print("-"*50)
```

```python
dataloader= DataLoader(dataset= data, batch_size= 1, shuffle= False)

for feature, label in dataloader:
    print(feature)
    print(label)
    print("-"*50)
```
