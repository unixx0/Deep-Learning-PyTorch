## Example 3
    - Finding the gradient of w and b of a perceptron
    - x is input and y is the actual output
    - We are using Sigmoid as activation function

```python
x= torch.tensor(6.9)
y= torch.tensor(0.0)


# we arent doing required grad = True becasue we donot want to calculate derivative with respect to x
```

```python
w= torch.tensor(1.0, requires_grad= True)
b= torch.tensor(0.0, requires_grad= True)

print(w)
print(b)
```

```python
# forward propagation
z= w*x + b
z
```

```python
y_pred= torch.sigmoid(z)
y_pred
```

```python
def binary_cross_entropy(y_pred, y_true):
    return (-y_true*torch.log(y_pred)- (1-y_true)*torch.log(1-y_pred))
```

```python
loss= binary_cross_entropy(y_pred, y)
loss
```

```python
## compute dy/dw
loss.backward()
w.grad
```

```python
b.grad
```

## Clearing grad
    - After each forward prop and backward we need to clear the grad because previous grad will be added 

```python
x= torch.tensor(data= 5.0, requires_grad=True)
x
```

```python
# without clearing grad
for i in range (5):
    y= x**2 + 2*x
    y.backward()
    x.grad
    print("y: ", y)
    print("x.grad: ", x.grad)
```

```python
x.grad.zero_()  # cleared previous grad
# with clearing grad
for i in range (5):
    y= x**2 + 2*x
    y.backward()
    x.grad
    print("y: ", y)
    print("x.grad: ", x.grad)
    x.grad.zero_()
```
