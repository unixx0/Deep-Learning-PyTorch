### 2. Reduction Operation

```python
e= torch.randint(size= (2, 3), low= 0, high= 10, dtype= torch.float64)
e
```

```python
# to find the sum of all the elements inside tensor
torch.sum(e)
```

```python
# sum along column
torch.sum(e, dim= 0, dtype= torch.int32) # dim = 1 for rows
```

```python
# mean
torch.mean(e)  #tensor should be a float for mean
```

```python
#mean along row
torch.mean(e, dim= 1)
```

```python
# for median
## we can also perform row wise or column wise as above
torch.median(e)

# similiarly .min, .max, .var, .cov, .std
```

```python
# argmax
## gives the position of largest item. Similiarly argmin for postion of smallest item
torch.argmax(e)
```

```python
torch.cov(e)
```

## Reshaping Tensors

```python
tensor= torch.ones(size= (3,2,1,4))
tensor.shape
```

```python
tensor
```

```python
# rehshape
tensor.reshape(6, 4)    #during rehsape the product of shapes must be same
```

```python
#flatten(converts to 1d)
tensor.flatten()
```

```python
# permute
permuted_tensor= tensor.permute(3, 2, 1, 0)
permuted_tensor.shape
```

```python
# unsqueeze
## adds dimension of 1 in desired postion
unsq_tensor= tensor.unsqueeze(1)
unsq_tensor
```

```python
unsq_tensor.shape
```
